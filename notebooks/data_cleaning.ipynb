## Standardising Cluster Identifiers
Cluster identiiferiers are formatted differently between catalouges. A common Cluster_ID column will be created to allow the datasets to be matched and merged. 

### VandenBerg Cluster Identifiers
Most clusters are identified by an NGC number. For clusters without an NGC number (#NGC = XXXX), the listed cluster name is used instead.

`[["#NGC", "Name"]]` - only shows me the columns called #NGC and Name.  
`.tail(10)` - shows me the last 10 rows.  

We only want the last 10 rows of the VandenBerg table as the unusal non-NGC cluster entries are near the bottom. 

```python
print(vdb[["#NGC", "Name"]].tail(10))
```

`vbd["Cluster_ID"] =` - creates a new column in the VandenBerg table called Cluster_ID
`np.where(condition, do_this_if_true, do_this_if_false)` - works like an if/else statement

`vdb["#NGC"].astype(str).str.strip() != "XXXX"` - Look at the #NGC value, treat it as text, remove any accidental spaces, and ask whether it is not equal to XXXX.

If thats true then, `"NGC" + vdb["#NGC"].astype(str).str.strip()`.  
If its false (meaning NGC = XXXX) then, `vdb["Name"].astype(str).str.replace(" ", "", regex=False)`.  

`.astype(str)` - convert data to str, a string. 
`.str.strip()` - removes spaces from the beginning and end of text
`.str.replace(" ", "", regex=False)` - removes spaces inside of text, " " gets replaced with "". The regex=False treats the thing we are replacing as normal text, not as a special pattern. 

```python
vdb["Cluster_ID"] = np.where(
    vdb["#NGC"].astype(str).str.strip() != "XXXX",
    "NGC" + vdb["#NGC"].astype(str).str.strip(),
    vdb["Name"].astype(str).str.replace(" ", "", regex=False)
)
```

```python
print(vdb[["#NGC", "Name", "Cluster_ID"]].tail(10))
```

### Harris Cluster Identifiers
Similarly, spaces were removed from the Harris catalogue identifiers to create a consistent Cluster_ID format for catalogue matching.

```python
harris1["Cluster_ID"] = (
    harris1["ID"]
    .astype(str)
    .str.strip()
    .str.replace(" ", "", regex=False)
)

harris3["Cluster_ID"] = (
    harris3["ID"]
    .astype(str)
    .str.strip()
    .str.replace(" ", "", regex=False)
)
```

Checking that this worked:

```python
print(harris1[["ID", "Cluster_ID"]].head())
print(harris1[["ID", "Cluster_ID"]].tail())
```

### Krause Cluster Identifiers
A standardised Cluster_ID column was also created for the Krause catalogue.

```python
krause["Cluster_ID"] = (
    krause["Object"]
    .astype(str)
    .str.strip()
    .str.replace(" ", "", regex=False)
)

print(krause[["Object", "Cluster_ID"]].head())
print(krause[["Object", "Cluster_ID"]].tail())
```

### Resolving Catalouge Naming Differences
Some clusters use different names or abbreviations between catalogues. Verified alternative names were mapped to a common identifier. Objects genuinely absent from another catalogue were not altered.

`"old value":"new value"`.  
For example, `"Ter8": "Terzan8"` means whenever we find Ter8, replace it with Terzan8.

```python
vdb_aliases = {
    "Ter8": "Terzan8"
}

krause_aliases = {
    "Palomar12": "Pal12",
    "Ruprecht106": "Rup106"
}

vdb["Cluster_ID"] = vdb["Cluster_ID"].replace(vdb_aliases)

krause["Cluster_ID"] = krause["Cluster_ID"].replace(krause_aliases)
```

Checking the sets again after changing the IDs: 

```python
vdb_ids = set(vdb["Cluster_ID"])
krause_ids = set(krause["Cluster_ID"])
harris1_ids = set(harris1["Cluster_ID"])

print("VandenBerg not in Harris:")
print(vdb_ids - harris1_ids)

print("\nVandenBerg not in Krause:")
print(vdb_ids - krause_ids)

print("\nKrause not in Harris:")
print(krause_ids - harris1_ids)
```

Arp21 in VandenBerg catalogue does not match the Harris catalogue and need to verify it against the original source. 

### Mannual Identifier Corrections
The original VandenBerg et al. (2013) Table 2 was checked to resolve catalogue naming discrepancies. The supplied Arp21 identifier corresponds to Arp 2, and Ter8 corresponds to Terzan 8 in the Harris catalogue naming convention.

```python
vdb_aliases = {
    "Arp21": "Arp2",
    "Ter8": "Terzan8"
}

vdb["Cluster_ID"] = vdb["Cluster_ID"].replace(vdb_aliases)

krause_aliases = {
    "Palomar12": "Pal12",
    "Ruprecht106": "Rup106"
}

krause["Cluster_ID"] = krause["Cluster_ID"].replace(krause_aliases)
```

Re-checking the matches: 

```python
vdb_ids = set(vdb["Cluster_ID"])
krause_ids = set(krause["Cluster_ID"])
harris1_ids = set(harris1["Cluster_ID"])
harris3_ids = set(harris3["Cluster_ID"])

print("VandenBerg not in Harris I:")
print(vdb_ids - harris1_ids)

print("\nVandenBerg not in Harris III:")
print(vdb_ids - harris3_ids)
```

Now every VandenBerg cluster has a matching Harris cluster. 

## Merging Catalogues
### Merging Harris Part I and Part III
`pd.merge(...)` - use pandas' table merging function.  
`on="Cluster_ID"` - use Cluster_ID to decide which rows belong together.  
`how=inner` - only keeps clusters that occour in both tables.  
`suffixes=("_pos","_kin")` - if both tables contain a column with the same name, pandas distinguishes them adding suffixes.  
`validate="one_to_one"` - tells pandas each cluster should appear once. If not, Python will not merge. 

```python
harris = pd.merge(
    harris1,
    harris3,
    on="Cluster_ID",
    how="inner",
    suffixes=("_pos", "_kin"),
    validate="one_to_one"
)
```

To check the merge worked, we can check the shape since both original Harris tables had 157 clusters we should still have 157 rows. 

```python
print(harris.shape)
```

```python
harris.head()
```

`.nunique()` - number of unique values.

```python
print(harris["Cluster_ID"].nunique())
```

### Merging VandenBerg onto Harris
As VandenBerg has 55 clusters compared to Harris 157 and the project is mainly based around the 55 VandenBerg clustres with age/metallicity measurements we can keep all the VandenBerg rows and add Harris whenever a matching cluster exists. 

`how=left` - keeps every row from the left-hand table (VandenBerg) and adds the matching information from the right-hand table wherever possible. 

```python
master = pd.merge(
    vdb,
    harris,
    on="Cluster_ID",
    how="left",
    validate="one_to_one"
)
```

Checking this:

```python
print(master.shape)
master.head()
```

```python
print(master["Cluster_ID"].nunique())
```

Check if the following variables are in the master: 

```python
master[
    ["Cluster_ID", "Age", "Age_err", "FeH", "R_gc", "v_r", "v_LSR"]
].head()
```

Check the missing values again but only in the important columns: 

```python
important_columns = [
    "Age",
    "Age_err",
    "FeH",
    "R_gc",
    "v_r",
    "v_r_e",
    "v_LSR",
    "sig_v",
    "sig_v_e"
]

master[important_columns].isna().sum()
```

All 55 VandenBerg clusters matched to the Harris catalogues. Age, metallicity, Galactocentric distance, radial velocity and LSR velocity are complete for the merged sample. Velocity dispersion measurements are unavailable for 18 clusters.

### Merging Krause onto master
Renaming krause column names common to master:

```python
krause = krause.rename(columns={
    "Age": "Age_Krause",
    "FeH": "FeH_Krause"
})
```

Merge Krause onto master while keeping VandenBerg as the primary dataset:

```python
master = pd.merge(
    master,
    krause[
        ["Cluster_ID", "Mstar", "rh", "C5", "Age_Krause", "FeH_Krause"]
    ],
    on="Cluster_ID",
    how="left",
    validate="one_to_one"
)
```

Check:

```python
print(master.shape)
print(master["Cluster_ID"].nunique())
```

```python
master[
    ["Mstar", "rh", "C5", "Age_Krause", "FeH_Krause"]
].isna().sum()
```

All 5 Krause varaibles are missing for exactly three rows, it suggests that those three clusteers aren't in the Krause catalogue rather than measurements randomly being missing. Finding out which clusters these are:

```python
master[
    master["Mstar"].isna()
][["Cluster_ID", "Age", "FeH"]]
```

Krause measurements were successfully matched to 52 of the 55 clusters in the primary VandenBerg sample. Three clusters are absent from the Krause catalogue and retain missing values for the Krause-derived mass, radius, compactness, age and metallicity parameters.
